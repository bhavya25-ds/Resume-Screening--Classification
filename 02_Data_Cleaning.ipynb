# 02 Data Cleaning

Producing a `clean_text` column: lowercase, strip URLs/HTML/punctucation/digits, remove stopwords and very short tokens. This is the text that TF-IDF vectorizer will consume in notebook 03.

## 1. Import libraries

```python
import pandas as pd
import Utility
```

```python
df= Utility.load_data()
print('rows: ', len(df))
```

## 2. Apply Cleaning

```python
df['clean_text']= df['text'].astype(str).map(Utility.clean_text)
df= df[df['clean_text'].str.strip() != '']
print('after cleaning: ', len(df), 'rows')
```

## 3. Before/After example

```python
ex=df.iloc[0]
print('RAW  :',str(ex['text'])[:200].replace(chr(10),' '))
print('\nCLEAN:',ex['clean_text'][:200])
```

## 3. Save Clean Data

```python
df[['clean_text', 'label']].to_csv('resume_cleaned.csv', index= False)
print(df.shape)
```
